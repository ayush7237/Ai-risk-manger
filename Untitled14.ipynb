```python
print(df.shape)
print(df['isFraud'].value_counts())
print(df['isFraud'].value_counts(normalize=True) * 100)  # fraud rate %
```

```python
# stratified subsample: keep ALL fraud cases + a random sample of non-fraud
fraud = df[df['isFraud'] == 1]
non_fraud = df[df['isFraud'] == 0].sample(n=300_000, random_state=42)

df_small = pd.concat([fraud, non_fraud]).sample(frac=1, random_state=42).reset_index(drop=True)
print(df_small.shape)
print(df_small['isFraud'].value_counts())
print(df[df['isFraud'] == 1]['type'].value_counts())
```

```python
# scope to the only transaction types where fraud occurs
df_scoped = df[df['type'].isin(['CASH_OUT', 'TRANSFER'])].copy()
print(df_scoped.shape)
print(df_scoped['isFraud'].value_counts(normalize=True) * 100)
```

```python
# 1. Did the sender's balance drop to exactly zero? (classic fraud signal - draining an account)
df_scoped['orig_balance_drained'] = (df_scoped['newbalanceOrig'] == 0).astype(int)

# 2. Error in sender's balance: expected new balance vs actual new balance
df_scoped['errorBalanceOrig'] = df_scoped['oldbalanceOrg'] - df_scoped['amount'] - df_scoped['newbalanceOrig']

# 3. Error in receiver's balance
df_scoped['errorBalanceDest'] = df_scoped['oldbalanceDest'] + df_scoped['amount'] - df_scoped['newbalanceDest']

# 4. Did the receiver's balance start at zero? (common in fraud - mule/fresh accounts)
df_scoped['dest_balance_was_zero'] = (df_scoped['oldbalanceDest'] == 0).astype(int)

# 5. Amount as a fraction of sender's balance (fraud often drains large %)
df_scoped['amount_to_balance_ratio'] = df_scoped['amount'] / (df_scoped['oldbalanceOrg'] + 1)  # +1 avoids divide-by-zero

df_scoped[['isFraud', 'orig_balance_drained', 'errorBalanceOrig', 'errorBalanceDest', 'dest_balance_was_zero', 'amount_to_balance_ratio']].groupby('isFraud').mean()
```

```python
df_scoped[['isFraud', 'amount_to_balance_ratio']].groupby('isFraud').median()
```

```python
# one-hot encode the 'type' column (only CASH_OUT/TRANSFER remain)
df_model = pd.get_dummies(df_scoped, columns=['type'], drop_first=True)

# drop columns that shouldn't be model inputs
df_model = df_model.drop(columns=['nameOrig', 'nameDest', 'isFlaggedFraud'])

print(df_model.columns.tolist())
```

```python
# time-aware split: train on earlier steps, test on later ones (simulates real deployment)
split_step = df_model['step'].quantile(0.8)
train = df_model[df_model['step'] <= split_step]
test = df_model[df_model['step'] > split_step]

print(f"Split at step {split_step}")
print(f"Train: {train.shape}, fraud rate: {train['isFraud'].mean()*100:.2f}%")
print(f"Test: {test.shape}, fraud rate: {test['isFraud'].mean()*100:.2f}%")

X_train = train.drop(columns=['isFraud'])
y_train = train['isFraud']
X_test = test.drop(columns=['isFraud'])
y_test = test['isFraud']
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, precision_recall_curve, average_precision_score

logreg = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
logreg.fit(X_train, y_train)

y_pred = logreg.predict(X_test)
y_proba = logreg.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred, target_names=['legit', 'fraud']))
print(f"PR-AUC: {average_precision_score(y_test, y_proba):.4f}")
```

```python
!pip install xgboost -q
from xgboost import XGBClassifier

scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgb = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=scale_pos_weight,
    eval_metric='aucpr',
    random_state=42
)
xgb.fit(X_train, y_train)

y_pred_xgb = xgb.predict(X_test)
y_proba_xgb = xgb.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred_xgb, target_names=['legit', 'fraud']))
print(f"PR-AUC: {average_precision_score(y_test, y_proba_xgb):.4f}")
```

```python
import pandas as pd
importances = pd.Series(xgb.feature_importances_, index=X_train.columns).sort_values(ascending=False)
print(importances)
```

```python
leaky_features = ['errorBalanceOrig', 'orig_balance_drained']

X_train_clean = X_train.drop(columns=leaky_features)
X_test_clean = X_test.drop(columns=leaky_features)

xgb_clean = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=scale_pos_weight,
    eval_metric='aucpr',
    random_state=42
)
xgb_clean.fit(X_train_clean, y_train)

y_pred_clean = xgb_clean.predict(X_test_clean)
y_proba_clean = xgb_clean.predict_proba(X_test_clean)[:, 1]

print(classification_report(y_test, y_pred_clean, target_names=['legit', 'fraud']))
print(f"PR-AUC: {average_precision_score(y_test, y_proba_clean):.4f}")

importances_clean = pd.Series(xgb_clean.feature_importances_, index=X_train_clean.columns).sort_values(ascending=False)
print(importances_clean)
```

```python
leaky_features_v2 = ['errorBalanceOrig', 'orig_balance_drained', 'newbalanceOrig', 'amount_to_balance_ratio']

X_train_v2 = X_train.drop(columns=leaky_features_v2)
X_test_v2 = X_test.drop(columns=leaky_features_v2)

xgb_v2 = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=scale_pos_weight,
    eval_metric='aucpr',
    random_state=42
)
xgb_v2.fit(X_train_v2, y_train)

y_pred_v2 = xgb_v2.predict(X_test_v2)
y_proba_v2 = xgb_v2.predict_proba(X_test_v2)[:, 1]

print(classification_report(y_test, y_pred_v2, target_names=['legit', 'fraud']))
print(f"PR-AUC: {average_precision_score(y_test, y_proba_v2):.4f}")

importances_v2 = pd.Series(xgb_v2.feature_importances_, index=X_train_v2.columns).sort_values(ascending=False)
print(importances_v2)
```

```python
import numpy as np
from sklearn.metrics import precision_recall_curve

precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba_v2)

for target_precision in [0.6, 0.7, 0.8]:
    idx = np.argmax(precisions >= target_precision)
    if idx < len(thresholds):
        print(f"Precision >= {target_precision}: threshold={thresholds[idx]:.3f}, recall={recalls[idx]:.3f}")
```

```python
chosen_threshold = 0.742
y_pred_final = (y_proba_v2 >= chosen_threshold).astype(int)

# get the actual transaction amounts for the test set
test_amounts = test['amount'].values

# true positives: fraud correctly caught -> money saved
tp_mask = (y_pred_final == 1) & (y_test == 1)
money_saved = test_amounts[tp_mask].sum()

# false positives: legit transactions wrongly flagged -> friction cost (not lost, just delayed/reviewed)
fp_mask = (y_pred_final == 1) & (y_test == 0)
money_flagged_legit = test_amounts[fp_mask].sum()
count_flagged_legit = fp_mask.sum()

# false negatives: fraud missed -> money lost
fn_mask = (y_pred_final == 0) & (y_test == 1)
money_missed = test_amounts[fn_mask].sum()

print(f"Fraud caught (money saved): {money_saved:,.0f}")
print(f"Fraud missed (money lost): {money_missed:,.0f}")
print(f"Legit transactions flagged (friction, not loss): {count_flagged_legit} totaling {money_flagged_legit:,.0f}")
```

```python
total_fraud_value = test_amounts[y_test == 1].sum()
recall_by_value = money_saved / total_fraud_value
print(f"% of fraud VALUE caught: {recall_by_value*100:.1f}%")

total_legit_value = test_amounts[y_test == 0].sum()
friction_rate = money_flagged_legit / total_legit_value
print(f"% of legit VALUE unnecessarily flagged: {friction_rate*100:.3f}%")
```
